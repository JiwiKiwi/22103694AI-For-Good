### 3.3 if / elif / else

*Slide 15 - 20 minutes*

Many AI systems make decisions with exactly the structure you are about to
write. The same logic that flags a "high-risk loan applicant" can reproduce
human bias.

Python checks the conditions from top to bottom and runs the first one that is
True. Indentation is not styling in Python - it is the syntax that decides
which lines belong to which branch.

- Create the variable `income` with `income = int(input('Annual income in euros: '))`
- Write an `if` / `elif` / `else` block that assigns a risk level:
  - `income` below 20000 gives `'high risk'`
  - `income` below 40000 gives `'medium risk'`
  - anything else gives `'low risk'`
- Print: `Based on income alone, this person is classified as: [risk level]`
- Then also ask for `age`, and combine the two with `and` or `or` so that the
  classification uses both.

````python
income = int(input('Annual income in euros: '))

# Classification based on income alone.
if income < 20000:
    income_risk = 'high risk'
elif income < 40000:
    income_risk = 'medium risk'
else:
    income_risk = 'low risk'

print('Based on income alone, this person is classified as: ' + income_risk)

# A second, illustrative classification that uses both income and age.
age = int(input('Age: '))

if income < 20000 or age < 18:
    combined_risk = 'high risk'
elif income < 40000 or age < 25:
    combined_risk = 'medium risk'
else:
    combined_risk = 'low risk'

print('Based on income and age, this person is classified as: ' + combined_risk)
````

## 4. Green or Not?

Everything from the three waves, combined into one program. Individual work. This one goes into your portfolio.

In this assignment you combine everything from today into one program: input,
type conversion, calculation, and a decision. This is a portfolio piece - polish
it at home if you do not finish it in class.

**Data:**

Return short-haul flight: 400 kg CO2 each
Driving: 0.23 kg CO2 per km
Meat-heavy diet: 1400 kg CO2 per year
Sustainable annual budget: 2500 kg CO2

**Classification thresholds:**

Below 2500 kg: under budget
2500 to 4000 kg: slightly over budget
Above 4000 kg: far over budget

**Assignment:**
- Make variables for the four reference values above.
- Ask the user how many return flights they take per year, how many kilometres
  they drive per year, and whether they eat meat daily. Convert each answer to
  the right type.
- Calculate the total annual CO2 and assign it to a variable.
- Use `if` / `elif` / `else` to classify the total against the thresholds.
- Print the type of one of your variables.
- Print a personalised message containing both the total and the classification,
  for example: `You emit 5150 kg CO2 per year. That is far over budget.`

````python
# Reference values.
short_haul_flight_co2 = 400
driving_co2_per_km = 0.23
meat_heavy_diet_co2 = 1400
sustainable_annual_budget = 2500

# Questions to the user and conversion to suitable data types.
return_flights = int(input('How many return flights do you take per year? '))
kilometres_driven = float(input('How many kilometres do you drive per year? '))
meat_answer = input('Do you eat meat daily? (yes/no): ').lower()
eats_meat_daily = meat_answer == 'yes'

# Calculate the annual CO2 total.
flight_emissions = return_flights * short_haul_flight_co2
driving_emissions = kilometres_driven * driving_co2_per_km
diet_emissions = meat_heavy_diet_co2 if eats_meat_daily else 0
total_annual_co2 = flight_emissions + driving_emissions + diet_emissions

# Classify the total.
if total_annual_co2 < sustainable_annual_budget:
    classification = 'under budget'
elif total_annual_co2 <= 4000:
    classification = 'slightly over budget'
else:
    classification = 'far over budget'

# Print a type and the result.
print(type(total_annual_co2))
print('You emit ' + str(round(total_annual_co2, 2)) + ' kg CO2 per year. That is ' + classification + '.')
````
